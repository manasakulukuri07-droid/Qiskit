```python
from qiskit import QuantumCircuit
from qiskit.primitives import StatevectorSampler
from qiskit.visualization import plot_histogram
import matplotlib.pyplot as plt
```

```python
# Create a circuit with exactly 2 qubits
qc_a = QuantumCircuit(2)

# Create a Bell state
qc_a.h(0)
qc_a.cx(0, 1)

# Measure both qubits
qc_a.measure_all()

# Display the circuit
print("Part A - Correlated Outcomes")
display(qc_a.draw("mpl"))
```

```python
sampler = StatevectorSampler()

result_a = sampler.run([qc_a], shots=100).result()

counts_a = result_a[0].data.meas.get_counts()

print("Part A Measurement Results:")
print(counts_a)

plot_histogram(counts_a, title="Part A - Correlated Outcomes")
plt.show()
```

```python
# Create a circuit with exactly 2 qubits
qc_b = QuantumCircuit(2)

# Create a Bell state
qc_b.h(0)
qc_b.cx(0, 1)

# Flip the second qubit
qc_b.x(1)

# Measure both qubits
qc_b.measure_all()

# Display the circuit
print("Part B - Opposite Outcomes")
display(qc_b.draw("mpl"))
```

```python
result_b = sampler.run([qc_b], shots=100).result()

counts_b = result_b[0].data.meas.get_counts()

print("Part B Measurement Results:")
print(counts_b)

plot_histogram(counts_b, title="Part B - Opposite Outcomes")
plt.show()
```

```python
print("Part A:", counts_a)
print("Part B:", counts_b)

plot_histogram(
    [counts_a, counts_b],
    legend=["Part A - Correlated", "Part B - Opposite"],
    title="Comparison of Quantum Correlations"
)

plt.show()
```
