```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
import matplotlib.pyplot as plt
```

Circuit 1 — Make 0 more frequent Prediction before running:

Gate: RY(π/3)
Probability of 0: 75%
Probability of 1: 25%
Expected result in 100 shots: approximately 75 zeros and 25 ones. The RY gate rotates the qubit by a specified angle, allowing us to control the measurement probabilities.

```python
import math

qc0 = QuantumCircuit(1, 1)

qc0.ry(math.pi / 3, 0)
qc0.measure(0, 0)

print(qc0.draw())
```

```python
simulator = AerSimulator()

job0 = simulator.run(qc0, shots=100)
result0 = job0.result()
counts0 = result0.get_counts()

print("Measurement results:", counts0)

plot_histogram(counts0)
plt.show()
```
